Ученику надо было найти произведение числа 136 на некоторое двузначное число, в котором цифра единиц
вдвое больше цифры десятков. По рассеянности он поменял местами цифры двузначного числа и получил
произведение на 1224 больше искомого. Чему равно искомое произведение? 

```python
from sympy import Eq
from sympy import solve
from sympy import symbols

x = symbols("x")
eq1 = Eq(136 * (10 * 2 * x + x) - 136 * (10 * x + 2 * x), 1224)
sol = solve([eq1], [x])
d = sol[x]
e = 2 * d
n = 10 * d + e
print(sol)
print(str(n) + " * 136 =", 136 * n)
```

```python

from sympy import Eq
from sympy import solve
from sympy import symbols

x, y = symbols("x y")
eq1 = Eq(x / y, 2)
eq2 = Eq(136 * (10 * x + y) - 136 * (10 * y + x), 1224)
sol = solve([eq1, eq2], [x, y])
d = sol[y]
e = sol[x]
n = 10 * d + e
print(sol)
print(str(n) + " * 136 =", 136 * n)
```
